```python
# Load dataset
import pandas as pd
```

```python
sales = pd.read_csv("/content/Sales Report .csv")
nomba = pd.read_csv("/content/Nomba.csv")
```

```python
# Preview the First Five Records of the Nomba Dataset.
nomba.head()
```

```python
#Convert Transaction Date Columns to Datetime Format
sales["transaction_date"] = pd.to_datetime(
    sales["transaction_date"],
    format="mixed"
)

nomba["transaction_date"] = pd.to_datetime(
    nomba["transaction_date"],
    format="mixed"
)
```

```python
#Preview the First Ten Transaction Dates from the Sales Report
print(sales["transaction_date"].head(10))
```

```python
#Standardize Payment Method Values
sales["payment Method"] = sales["payment Method"].replace({
    "POS (Nomba)": "purchase",
    "Transfer (Nomba)": "pay_by_transfer"
})
```

```python
#Standardize Payment Method Values
sales["payment Method"] = sales["payment Method"].replace({
    "POS (Nomba)": "purchase",
    "Transfer (Nomba)": "pay_by_transfer"
})
```

```python
#Display All Column Names in the Nomba Dataset
print(nomba.columns.tolist())
```

```python
 #Remove Leading and Trailing Spaces from Nomba Column Names
nomba.columns = nomba.columns.str.strip()
```

```python
# Display the List of Column Names in the Nomba Dataset
print(nomba.columns.tolist())
```

```python
#Convert Amount Columns to Numeric Data Type
sales["amount"] = pd.to_numeric(
    sales["amount"],
    errors="coerce"
)

nomba["transaction Amount"] = pd.to_numeric(
    nomba["transaction Amount"],
    errors="coerce"
)
```

```python
#Sort Transactions by Transaction Date
sales = sales.sort_values("transaction_date")

nomba = nomba.sort_values("transaction_date")
```

```python
#Merge the Sales Report and Nomba Datasets Based on Transaction Amount
merged = sales.merge(
    nomba,
    left_on="amount",
    right_on="transaction Amount",
    suffixes=("_sales", "_nomba")
)
```

```python
#Filter Transactions with Matching Payment Methods
merged = merged[
    merged["payment Method"]
    ==
    merged["transactionType"]
]
```

```python
#Calculate the Time Difference Between Sales and Nomba Transactions
merged["time_difference"] = (
    merged["transaction_date_sales"]
    -
    merged["transaction_date_nomba"]
).abs()
```

```python
#Preview the Calculated Time Differences
matched = merged[
    merged["time_difference"]
    <=
    pd.Timedelta(hours=1)
]
```

```python
#Create an Independent Copy and Assign Reconciliation Status
matched = matched.copy()

matched["status"] = "Matched"
```

```python
 #Assign the Reconciliation Status to Matched Transactions
matched["status"] = "Matched"
```

```python
 #Identify Unmatched Sales Transactions
unmatched_sales = sales[
    ~sales.index.isin(
        matched.index
    )
]
```

```python
# Identify Unmatched Nomba Transactions
unmatched_nomba = nomba[
    ~nomba.index.isin(
        matched.index
    )
]
```

```python
#Export Reconciliation Reports to CSV Files
matched.to_csv(
    "Matched_Transactions.csv",
    index=False
)

unmatched_sales.to_csv(
    "Unmatched_Sales.csv",
    index=False
)

unmatched_nomba.to_csv(
    "Unmatched_Nomba.csv",
    index=False
)
```
